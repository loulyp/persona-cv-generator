```python
from datasets import load_dataset
import json

# Load the dataset (using the 'persona' subset)
dataset = load_dataset("proj-persona/PersonaHub", "persona", split="train")

# Filter for Arabic/Muslim keywords
keywords = ["arabic", "saudi", "egypt", "emirati", "muslim", "islamic", "qatar", "gcc" ,"middle eastern"]
filtered_personas = [
    p['persona'] for p in dataset
    if any(k in p['persona'].lower() for k in keywords)
]
```

```python
import pandas as pd

discovered_personas = []
count = 0
limit = 10 # Let's find the first 10 for examination

for entry in dataset:
    text = entry['persona'].lower()
    if any(k in text for k in keywords):
        discovered_personas.append(entry['persona'])
        count += 1
    if count >= limit:
        break

# Display as a clean table
df = pd.DataFrame(discovered_personas, columns=["Persona Description"])
print(count)
df.head(20)
```
